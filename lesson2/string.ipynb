```python
string = "Hello, World!"
string2 = 'Python is fun!'
string3 = """This is a
    multi-line
string."""
print(string)
print(string2)
print(string3)
```

```python
mystring = "Hello, this is my string!"
print(mystring)
print(mystring[0]) #first
print(mystring[-1]) #last
print(len(mystring)) #length
print(mystring[3:8]) #substring from index 3 to 7
print(mystring[-10:-2]) #substring from -10 to -3

#reverse
print(mystring[::-1]) #reversed string
```

```python
mystring = "Hello, this is my string!"
print(mystring.upper())
print(mystring.lower())
print(mystring.replace('this', 'that'))
print(mystring.split())
```

```python
price = 10
item = 'notebook'

# f-string with format
print(f'{item} costs {price:.2f}$.')
```
